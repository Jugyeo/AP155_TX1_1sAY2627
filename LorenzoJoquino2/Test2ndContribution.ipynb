```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def plotFunction(x, y, ylabel, filepath):
    fig, ax = plt.subplots(figsize = (6,6))
    ax.scatter(x,y)
    ax.set_ylabel(ylabel)
    ax.set_xlabel('x')
    fig.tight_layout()
    fig.show()
    fig.savefig(filepath)
    return 
```

```python
x = np.linspace(-5,5,100)
y = np.sin(x) + x
ylabel = r'$\sin(x) +x$'
filepath = "sampleImage.jpg"
plotFunction(x, y, ylabel, filepath)
```
